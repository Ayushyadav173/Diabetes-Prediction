# Diabetes Prediction using DT SVC 

```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import BernoulliNB

import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score , confusion_matrix
import seaborn as sns 


```

```python

#read the data file
data = pd.read_csv(r"D:\up ayush counselling\Diabetes-Prediction.-main\Diabetes-Prediction.-main\Dataset\diabetes.csv")
data.head()
```

```python
data.describe()
```

```python
#here few misconception is there lke BMI can not be zero, BP can't be zero, glucose, insuline can't be zero so lets try to fix it
# now replacing zero values with the mean of the column
data['BMI'] = data['BMI'].replace(0,data['BMI'].mean())
data['BloodPressure'] = data['BloodPressure'].replace(0,data['BloodPressure'].mean())
data['Glucose'] = data['Glucose'].replace(0,data['Glucose'].mean())
data['Insulin'] = data['Insulin'].replace(0,data['Insulin'].mean())
data['SkinThickness'] = data['SkinThickness'].replace(0,data['SkinThickness'].mean())

```

```python
data.describe()
```

```python
#to check for outliers 
fig,ax = plt.subplots(figsize= (15,10))
sns.boxplot(data=data, width=0.5,ax=ax ,fliersize=3)
```

```python
X=data.drop(columns=['Outcome'])
y=data['Outcome']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=0)
X_train.shape,X_test.shape
```

```python
import pickle

def scaler_standard(X_train,X_test):
    #scaling the data 
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    #saving the model
    file = open(r'D:\up ayush counselling\Diabetes-Prediction.-main\Diabetes-Prediction.-main\Dataset\diabetes.csv','wb')
    pickle.dump(scaler,file)
    file.close()
    
    return X_train_scaled, X_test_scaled
```

```python
X_train_scaled , X_test_scaled = scaler_standard(X_train,X_test)
```

```python
X_train_scaled
```

```python
# decision Tree model Trainings
import warnings 
warnings.filterwarnings('ignore')
```

```python
parameter = {
    'criterion' : ['gini','entropy','log_loss'],
    'splitter'  : ['best','random'],
    'max_depth' : [1,2,3,4,5],
    'max_features' : ['auto','sqrt','log2']

 }
```

```python
from sklearn.model_selection import GridSearchCV
classifier = DecisionTreeClassifier()
clf = GridSearchCV(classifier,param_grid=parameter,cv=5,scoring='accuracy',verbose=3)
clf.fit(X_train,y_train)
```

```python
clf.best_params_
```

```python
classifier = DecisionTreeClassifier(criterion='entropy',max_depth=5,max_features='log2',splitter='random')
classifier.fit(X_train,y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
```

```python
conf_mat = confusion_matrix(y_test,y_pred)
conf_mat
```

```python
true_positive = conf_mat[0][0]
false_positive = conf_mat[0][1]
false_negative = conf_mat[1][0]
true_negative = conf_mat[1][1]
```

```python
Accuracy = (true_positive+true_negative)/(true_positive+false_negative+false_positive+true_negative)
```

```python
Accuracy
```

```python
precision = true_positive/(true_positive+false_positive)
```

```python
precision
```

```python
Recall = true_positive/(true_positive+false_negative)
Recall
```

```python
F1_score = 2*(Recall*precision)/(Recall+precision)
F1_score
```

```python
import pickle
file = open(r'D:\up ayush counselling\Diabetes-Prediction.-main\Diabetes-Prediction.-main\Dataset\diabetes.csv','wb')
pickle.dump(classifier,file)
file.close()
```
